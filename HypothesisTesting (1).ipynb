## Hypothesis Testing


# 1. Hypothesis Definitions

Here I formally define the hypotheses for each research question.
The goal is to test whether daily time spent on different platforms is statistically associated with my anxiety, social comparison, productivity, and motivation scores.

RQ1 – Platform Use and Anxiety

H₀₁ (Null): Social media usage (Instagram, Twitter, TikTok) is not related to daily anxiety levels.

H₁₁ (Alternative): Time spent on platforms—especially Twitter—is positively associated with anxiety.

RQ2 – Platform Use and Social Comparison

H₀₂ (Null): There is no relationship between platform usage and social comparison.

H₁₂ (Alternative): Instagram and TikTok usage are positively associated with social comparison.

RQ3 – Platform Use, Productivity, and Motivation

H₀₃ (Null): Platform usage has no relationship with productivity or motivation.

H₁₃ (Alternative): YouTube may relate positively to productivity/motivation, while TikTok may relate negatively.

These hypotheses guide the statistical analysis that follows.

## 2. Loading the Dataset
Below I load the dataset I prepared during the data collection phase.

```python
import pandas as pd
from scipy.stats import pearsonr

df = pd.read_csv("daily_social_media.csv")

df.head()
```

## 3. Visual Inspection Before Statistical Tests

Before applying Pearson correlation, it is useful to visually check whether a linear relationship seems plausible.
Here I generate scatterplots for each platform relative to anxiety.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
sns.scatterplot(x=df["ig_min"], y=df["anxiety"])
plt.title("Instagram vs Anxiety")

plt.subplot(1,3,2)
sns.scatterplot(x=df["tw_min"], y=df["anxiety"])
plt.title("Twitter vs Anxiety")

plt.subplot(1,3,3)
sns.scatterplot(x=df["tt_min"], y=df["anxiety"])
plt.title("TikTok vs Anxiety")

plt.tight_layout()
plt.show()
```

# 4. Pearson Correlation Tests
## 4.1 Platform Use and Anxiety

The following code computes Pearson correlation coefficients and p-values for the relationship between platform usage and anxiety.

```python
anx_vars = ["ig_min", "tw_min", "tt_min"]

for col in anx_vars:
    r, p = pearsonr(df[col], df["anxiety"])
    print(f"{col} → Anxiety: r = {r:.3f}, p = {p:.3f}")
```

## 4.2 Platform Use and Social Comparison

```python
soc_vars = ["ig_min", "tw_min", "tt_min"]

for col in soc_vars:
    r, p = pearsonr(df[col], df["soc_comp"])
    print(f"{col} → Social Comparison: r = {r:.3f}, p = {p:.3f}")
```

## 4.3 Platform Use, Productivity, and Motivation

```python
prod_mot_vars = ["yt_min", "tt_min", "ig_min"]

for col in prod_mot_vars:
    r1, p1 = pearsonr(df[col], df["productivity"])
    r2, p2 = pearsonr(df[col], df["motivation"])
    print(f"{col} → Productivity: r = {r1:.3f}, p = {p1:.3f}")
    print(f"{col} → Motivation:   r = {r2:.3f}, p = {p2:.3f}")
    print()
```

## 5. Interpretation of Results

Across all three research questions, the results indicate that the relationships between daily platform use and my mood or productivity measures are weak overall.

For RQ1 (Anxiety), Instagram, Twitter, and TikTok showed small positive correlations with anxiety, but none were strong enough to suggest a meaningful statistical relationship. The direction fits the hypothesis, but the evidence is limited.

For RQ2 (Social Comparison), Instagram and TikTok had slightly higher correlations with social comparison than Twitter, which aligns with expectations. However, the associations were still modest and should be interpreted cautiously.

For RQ3 (Productivity & Motivation), YouTube showed a small positive relationship with productivity and motivation, while TikTok showed a slight negative trend. These patterns match the hypotheses but remain weak in magnitude.

Overall, the findings suggest subtle trends rather than strong effects, likely due to the small sample size and daily variability. While the results point in the expected directions, the correlations are not strong enough to draw firm conclusions. Additional data would be needed to confirm these patterns.
